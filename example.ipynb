```python
def trace(n: int) -> int:
    print(f"호출: trace({n})")
    if n <= 1:
        print(f" 기저: return 1")
        return 1
    result: int = n + trace(n - 2)
    print(f" 반환: trace({n}) = {result}")
    return result

trace(5)
```

```python
def merge(left: list, right: list) -> list:
    result = []
    i, j = 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

```python
def merge_sort(lst: list[int]) -> list[int]:
    if len(lst) <= 1:
        return lst
    mid: int = len(lst) // 2
    left: list[int] = merge_sort(lst[:mid])
    right: list[int] = merge_sort(lst[mid:])
    return merge(left , right)
merge_sort([5, 2, 8, 1, 9, 3])
```

```python
def mystery(a: int, b: int) -> int:
    if b == 0:
        return a
    return mystery(b, a % b)
mystery(48,18)
```
