```python
import numpy as np
import pandas as pd
works = pd.DataFrame({
'작가': ['김유정', '김유정', '김유정',
'현진건', '현진건',
'이상', '이상',
'나도향', '나도향',
'채만식', '이효석'],
'작품': ['봄봄', '동백꽃', '만무방',
'운수 좋은 날', 'B사감과 러브레터',
'날개', '권태',
'벙어리 삼룡이', '물레방아',
'레디메이드 인생', '메밀꽃 필 무렵'],
'연도': [1935, 1936, 1934,
1924, 1925,
1936, 1937,
1925, 1925,
1934, np.nan], # 이효석 '메밀꽃 필 무렵' 발표연도가 누락됨
'글자수': [12500, 9800, 13600,
11200, 7400,
18300, 14100,
16800, 14500,
22000, 11800],
})
authors = pd.DataFrame({
'작가': ['김유정', '현진건', '이상', '나도향', '채만식', '이효석', '염상섭'],
'생년': [1908, 1900, 1910, 1902, 1902, 1907, 1897],
'몰년': [1937, 1943, 1937, 1926, 1950, 1942, 1963],
})
```

```python
#Q1 (a)
print(works.shape)
print(works.dtypes)
print(works.describe())

#Q1 (b)
works.isna().sum()
```

```python
#Q1 (c)
works2 = works.fillna(1936)
works2['연도'] = works2['연도'].astype(int) 
print(works2.dtypes)
print(works2.tail(3))
```

Q1 
(a) 
데이터의 크기와 타입, 요약 통계를 확인한다.
관찰: 연도와 글자수는 모두 정수 데이터지만 연도 열에는 결측치가 있다. NaN은 정수형에는 없는 표현으로, NaN이 섞이면 정수 열도 float 처리된다.


(b) 
'연도'에 1건의 결측치가 있다.
isna() 메소드로 각 원소의 결측 여부를 판단하고, .sum()을 연결하여 각 열별 결측치 개수의 합계를 구한다.

Q1 (c)
fillna() 메소드로 연도 열의 결측치를 1936으로 대체하고, astype(int) 메소드를 사용해 해당 열의 자료형을 정수형으로 변환한다.


```python
#Q2 (a)
after_1930 = works2[works2['연도'] >= 1930]
kim_or_lee = works2[works2['작가'].isin(['김유정', '이상'])]
print(after_1930)
print(kim_or_lee)
```

```python
#Q2 (b)
def categorize(n: int) -> str:
    if n < 10000:
        return '짧음'
    elif 10000 <= n <= 15000:
        return '보통'
    else: 
        return '긴'

works2['분량'] = works2['글자수'].apply(categorize)
print(works2[['작품', '글자수', '분량']])
```

```python
#Q2 (c)
sorted_works2 = works2.sort_values(by=['연도', '글자수'], ascending=[True, False])
print(sorted_works2)
```

Q2
(a)
비교 연산자와 isin() 메소드로 조건에 맞는 행을 판별하는 불 시리즈를 생성한 뒤, 이를 활용해 부분 데이터프레임을 추출하는 불 인덱싱을 수행한다.

(b)
글자수에 따라 분량 등급을 문자열로 반환하는 파이썬 함수를 정의하고, apply() 메소드를 사용하여 시리즈의 모든 원소에 함수를 적용한 파생 열을 생성한다.

(c)
sort_values() 메소드의 by 인자에 기준 열 리스트를, ascending 인자에 각 열의 정렬 방식(오름차순/내림차순)을 지정하는 불 리스트를 넘겨 데이터를 재정렬한다.

```python
#Q3 (a)
author_stats = works2.groupby('작가')['글자수'].agg(['mean', 'count', 'sum'])
print(author_stats)
print()

#(b)
volume_counts = works2['분량'].value_counts()
print(volume_counts)
print()

#(c)
author_size = works2.groupby(['작가', '분량']).size()
print(author_size)
```

Q3
(a)
groupby() 메소드로 데이터를 작가별로 그룹화한 뒤, agg() 메소드에 집계 함수 이름 리스트를 넘겨 평균, 개수, 총합을 한 번에 계산한다.

(b)
시리즈 객체에 value_counts() 메소드를 사용하여 각 분량 등급별 데이터의 등장 빈도(개수)를 집계한다.

(c)
(김유정, 보통)으로 총 2개의 작품이 해당하며, 이는 김유정 작가가 10000자에서 15000자 사이의 보통 분량의 소설을 주로 집필하는 경향이 있음을 나타낸다.

코드 설명
groupby()에 두 개의 기준 열을 리스트로 지정하여 그룹화하고, .size() 메소드를 호출해 각 조합별 행의 개수를 집계한다.

```python
#Q4 (a)
full = pd.merge(works2, authors, on='작가', how='left')

print(full)
print(full.shape)
```

```python
#Q4 (b)
full['집필 당시 나이'] = full['연도'] - full['생년']
print(full[['작가', '작품', '연도', '생년', '집필 당시 나이']])
```

```python
#Q4 (c)
avg_age = full.groupby('작가')['집필 당시 나이'].mean().sort_values()
print(avg_age)
```

Q4
(a)
pd.merge() 함수에 how='left'와 on='작가'를 지정해 왼쪽 표를 기준으로 데이터를 병합하며, 오른쪽 표에만 존재하는 작가는 공통 짝이 없어 최종 결과에서 제외된다.

(b)
'연도' 열과 '생년' 열을 빼는 코드를 한 줄로 작성하여, for 루프 없이 모든 행의 원소별 연산을 한꺼번에 처리하는 벡터화 연산을 수행한다.

(c)
병합된 데이터프레임을 작가별로 그룹화하여 '집필 당시 나이'의 평균을 구한 뒤, sort_values() 메소드를 연쇄 호출해 평균 나이 기준 오름차순으로 정렬한다.
